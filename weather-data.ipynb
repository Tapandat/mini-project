```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/kaggle/input/datasets/tapanfds/weather/india_2000_2024_daily_weather.csv")

df.columns = df.columns.str.strip()

print(df.head())
print(df.info())
print(df.describe())

print(df.columns)

date_col = [col for col in df.columns if "date" in col.lower()][0]

df[date_col] = pd.to_datetime(df[date_col])

numeric_cols = df.select_dtypes(include='number').columns

plt.figure(figsize=(10,5))
plt.plot(df[date_col], df[numeric_cols[0]])
plt.title("Line Plot")
plt.xlabel("Date")
plt.ylabel(numeric_cols[0])
plt.show()

plt.figure(figsize=(8,5))
plt.hist(df[numeric_cols[0]], bins=30)
plt.title("Histogram")
plt.xlabel(numeric_cols[0])
plt.ylabel("Frequency")
plt.show()

plt.figure(figsize=(8,5))
plt.scatter(df[numeric_cols[0]], df[numeric_cols[1]])
plt.title("Scatter Plot")
plt.xlabel(numeric_cols[0])
plt.ylabel(numeric_cols[1])
plt.show()

plt.figure(figsize=(8,5))
plt.boxplot(df[numeric_cols[0]].dropna())
plt.title("Box Plot")
plt.show()

plt.figure(figsize=(8,5))
top_values = df[numeric_cols[0]].head(10)
plt.bar(range(len(top_values)), top_values)
plt.title("Bar Plot")
plt.show()

plt.figure(figsize=(8,8))
pie_data = df[numeric_cols[0]].head(5)
plt.pie(pie_data, labels=range(1,6), autopct='%1.1f%%')
plt.title("Pie Chart")
plt.show()

plt.figure(figsize=(8,5))
plt.violinplot(df[numeric_cols[0]].dropna())
plt.title("Violin Plot")
plt.show()

plt.figure(figsize=(8,5))
plt.stem(df[numeric_cols[0]].head(20))
plt.title("Stem Plot")
plt.show()

plt.figure(figsize=(8,5))
plt.step(range(len(df.head(50))), df[numeric_cols[0]].head(50))
plt.title("Step Plot")
plt.show()

plt.figure(figsize=(8,5))
plt.fill_between(range(len(df.head(50))), df[numeric_cols[0]].head(50))
plt.title("Area Plot")
plt.show()

plt.figure(figsize=(10,6))
corr = df[numeric_cols].corr()
plt.imshow(corr, cmap='coolwarm')
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title("Heatmap")
plt.show()

plt.figure(figsize=(8,5))
plt.hexbin(df[numeric_cols[0]], df[numeric_cols[1]], gridsize=25)
plt.colorbar()
plt.title("Hexbin Plot")
plt.xlabel(numeric_cols[0])
plt.ylabel(numeric_cols[1])
plt.show()

plt.figure(figsize=(8,5))
plt.eventplot(df[numeric_cols[0]].head(50))
plt.title("Event Plot")
plt.show()

plt.figure(figsize=(8,5))
plt.stackplot(range(50),
              df[numeric_cols[0]].head(50),
              df[numeric_cols[1]].head(50))
plt.title("Stack Plot")
plt.show()

plt.figure(figsize=(8,5))
plt.acorr(df[numeric_cols[0]].dropna()[:100], maxlags=20)
plt.title("Autocorrelation Plot")
plt.show()

plt.figure(figsize=(8,5))
plt.xcorr(df[numeric_cols[0]].dropna()[:100],
          df[numeric_cols[1]].dropna()[:100],
          maxlags=20)
plt.title("Cross Correlation Plot")
plt.show()

plt.figure(figsize=(8,5))
plt.magnitude_spectrum(df[numeric_cols[0]].dropna()[:256])
plt.title("Magnitude Spectrum")
plt.show()

plt.figure(figsize=(8,5))
plt.psd(df[numeric_cols[0]].dropna()[:256])
plt.title("Power Spectral Density")
plt.show()

plt.figure(figsize=(8,5))
plt.specgram(df[numeric_cols[0]].dropna()[:256])
plt.title("Spectrogram")
plt.show()
```
